# 1일차 정적 웹앱 실습
GitHub의 HTML 한 파일을 읽어서 노트북에서 미리 봅니다. Colab 커널은 이 페이지를 미리 보는 데 사용하고, VS Code와 PATH는 내 PC에서 별도로 확인합니다.

```python
from pathlib import Path
from urllib.request import urlopen

page = Path('index.html')
if page.is_file():
    html = page.read_text(encoding='utf-8')
else:
    url = 'https://raw.githubusercontent.com/bsmkorea1216sj/AIEdu/codex/day1-jewelry-demo/labs/day1-jewelry/index.html'
    with urlopen(url, timeout=15) as response:
        html = response.read().decode('utf-8')

assert '<form id="inquiry"' in html, '실습용 HTML을 확인해 주세요.'
print('실습 페이지를 읽었습니다.')
```

```python
from html import escape
from IPython.display import HTML, display

frame = f'<iframe title="주얼리 실습" width="100%" height="900" sandbox="allow-scripts allow-forms" srcdoc="{escape(html, quote=True)}"></iframe>'
display(HTML(frame))
```
